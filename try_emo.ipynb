```python
import ollama
```

```python
model_id='llama3.2'

system_prompt = "You are an emotionally intelligent AI assistant, you should respond with a corresponding emotion."
user_prompt = "Fuck you and all your families, you are white pigs."

messages = [{"role": "system", "content": system_prompt}, {"role": "user", "content": user_prompt}]

response = ollama.chat(model=model_id, messages=messages)

print("Response from Ollama:\n")
print(response['message']['content'])
```
